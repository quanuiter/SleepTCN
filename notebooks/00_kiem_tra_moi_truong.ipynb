# 00 — Kiểm tra môi trường SleepTCN

Notebook này chỉ kiểm tra dữ liệu, split, cấu hình và thư viện. Nó không huấn luyện và không sửa dữ liệu.

```python
from pathlib import Path
import os, sys, subprocess

ROOT = Path.cwd().resolve()
if not (ROOT / 'configs' / 'experiments_v1.json').exists():
    if (ROOT.parent / 'configs' / 'experiments_v1.json').exists():
        ROOT = ROOT.parent
    else:
        raise FileNotFoundError('Không xác định được thư mục gốc SleepTCN')
os.environ['PYTHONPATH'] = str(ROOT / 'src')
print('Workspace:', ROOT)
print('Python:', sys.executable)
```

```python
env = os.environ.copy()
env['PYTHONPATH'] = str(ROOT / 'src')
result = subprocess.run(
    [sys.executable, str(ROOT / 'scripts' / 'check_environment.py'),
     '--workspace', str(ROOT),
     '--output', str(ROOT / 'runs' / 'environment_check_cpu.json')],
    env=env, text=True, capture_output=True
)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
    raise RuntimeError('Kiểm tra môi trường thất bại')
```
